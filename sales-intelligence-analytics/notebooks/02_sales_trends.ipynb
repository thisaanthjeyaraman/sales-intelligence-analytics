```
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
```

```
PROJECT_DIR = Path(__file__).resolve().parent.parent
DATA_PATH = PROJECT_DIR / "data" / "Sample - Superstore.csv"
OUTPUT_DIR = PROJECT_DIR / "outputs"
```

```
def main():
    df = pd.read_csv(DATA_PATH)
    df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="raise")

    # Initial analysis: all recorded order statuses.
    # We will create status-specific business KPIs later.
    df["Month"] = df["Order_Date"].dt.to_period("M")

    monthly_sales = (
        df.groupby("Month")["Net_Sales"]
        .sum()
        .sort_index()
    )

    print("MONTHLY NET SALES")
    print(monthly_sales.round(2).to_string())

    top_10_months = monthly_sales.nlargest(10)

    print("\nTOP 10 MONTHS BY NET SALES")
    print(top_10_months.round(2).to_string())

    OUTPUT_DIR.mkdir(exist_ok=True)

    plt.figure(figsize=(12, 6))
    monthly_sales.plot(kind="line", marker="o")

    plt.title("Monthly Net Sales Trend (2023–2025)")
    plt.xlabel("Month")
    plt.ylabel("Net Sales")
    plt.xticks(rotation=45)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()

    chart_path = OUTPUT_DIR / "monthly_net_sales_trend.png"
    plt.savefig(chart_path, dpi=300)
    plt.close()

    print(f"\nChart saved to: {chart_path}")
    print("Monthly sales analysis completed successfully.")
```

```
if __name__ == "__main__":
    main()
```
