```
from pathlib import Path
import pandas as pd
```

```
# Find the project directory
PROJECT_DIR = Path(__file__).resolve().parent.parent
DATA_PATH = PROJECT_DIR / "data" / "Sample - Superstore.csv"
```

```
def main():
    # Load the new e-commerce dataset
    df = pd.read_csv(DATA_PATH)

    # Convert order dates
    df["Order_Date"] = pd.to_datetime(
        df["Order_Date"], errors="raise"
    )

    # Display first five records
    print("FIRST FIVE E-COMMERCE RECORDS")
    print(df.head().to_string(index=False))

    # Dataset dimensions
    print("\nDATASET SIZE")
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])

    # Column information
    print("\nCOLUMN INFORMATION")
    df.info()

    # Missing values
    print("\nMISSING VALUES")
    print(df.isnull().sum().to_string())

    # Duplicate records
    print("\nDUPLICATE ROWS:", df.duplicated().sum())
    print(
        "DUPLICATE ORDER IDs:",
        df["Order_ID"].duplicated().sum()
    )

    # Business summary
    print("\nBUSINESS SUMMARY")
    print("Total Gross Sales:", round(df["Gross_Sales"].sum(), 2))
    print("Total Discounts:", round(df["Discount_Amount"].sum(), 2))
    print("Total Net Sales:", round(df["Net_Sales"].sum(), 2))
    print("Total Cost:", round(df["Cost"].sum(), 2))
    print("Total Recorded Profit:", round(df["Profit"].sum(), 2))
    print("Total Quantity:", df["Quantity"].sum())

    # Date range
    print("\nORDER DATE RANGE")
    print("First order:", df["Order_Date"].min().date())
    print("Last order:", df["Order_Date"].max().date())

    # Order status summary
    print("\nORDER STATUS COUNTS")
    print(df["Order_Status"].value_counts().to_string())

    print("\nData overview completed successfully.")
```

```
if __name__ == "__main__":
    main()
```
