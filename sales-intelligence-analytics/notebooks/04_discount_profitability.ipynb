```
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
```

```
PROJECT_DIR = Path(__file__).resolve().parent.parent
DATA_PATH = PROJECT_DIR / "data" / "Sample - Superstore.csv"
OUTPUT_DIR = PROJECT_DIR / "outputs"
```

```
def main():
    df = pd.read_csv(DATA_PATH)
    OUTPUT_DIR.mkdir(exist_ok=True)

    # Group orders by the discount percentage recorded in the dataset.
    discount_analysis = (
        df.groupby("Discount_Percentage")
        .agg(
            Order_Count=("Order_ID", "nunique"),
            Gross_Sales=("Gross_Sales", "sum"),
            Net_Sales=("Net_Sales", "sum"),
            Recorded_Profit=("Profit", "sum"),
        )
        .sort_index()
    )

    print("DISCOUNT AND PROFITABILITY ANALYSIS")
    print(discount_analysis.round(2).to_string())

    # Save the analysis for future use in SQL or Power BI.
    discount_analysis.to_csv(
        OUTPUT_DIR / "discount_profitability_analysis.csv"
    )

    # Chart 1: Recorded profit by discount percentage.
    plt.figure(figsize=(9, 5))
    discount_analysis["Recorded_Profit"].plot(kind="bar")
    plt.title("Recorded Profit by Discount Percentage")
    plt.xlabel("Discount Percentage")
    plt.ylabel("Recorded Profit")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.savefig(
        OUTPUT_DIR / "profit_by_discount.png",
        dpi=300
    )
    plt.close()

    # Chart 2: Net sales by discount percentage.
    plt.figure(figsize=(9, 5))
    discount_analysis["Net_Sales"].plot(kind="bar")
    plt.title("Net Sales by Discount Percentage")
    plt.xlabel("Discount Percentage")
    plt.ylabel("Net Sales")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.savefig(
        OUTPUT_DIR / "net_sales_by_discount.png",
        dpi=300
    )
    plt.close()

    print("\nAnalysis CSV and charts saved in the outputs folder.")
    print(
        "Note: These totals include all order statuses; "
        "they are not verified realized revenue or profit."
    )
```

```
if __name__ == "__main__":
    main()
```
