```
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
```

```
PROJECT_DIR = Path(__file__).resolve().parent.parent
DATA_PATH = PROJECT_DIR / "data" / "Sample - Superstore.csv"
OUTPUT_DIR = PROJECT_DIR / "outputs"
```

```
def main():
    df = pd.read_csv(DATA_PATH)
    OUTPUT_DIR.mkdir(exist_ok=True)

    # Initial analysis across all recorded order statuses.
    # Status-specific KPIs will be defined separately.

    # 1. Profit by category
    category_profit = (
        df.groupby("Category")["Profit"]
        .sum()
        .sort_values(ascending=False)
    )

    print("PROFIT BY CATEGORY")
    print(category_profit.round(2).to_string())

    plt.figure(figsize=(9, 5))
    category_profit.plot(kind="bar")
    plt.title("Recorded Profit by Product Category")
    plt.xlabel("Category")
    plt.ylabel("Profit")
    plt.xticks(rotation=30, ha="right")
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "profit_by_category.png", dpi=300)
    plt.close()

    # 2. Profit by region
    region_profit = (
        df.groupby("Region")["Profit"]
        .sum()
        .sort_values(ascending=False)
    )

    print("\nPROFIT BY REGION")
    print(region_profit.round(2).to_string())

    plt.figure(figsize=(9, 5))
    region_profit.plot(kind="bar")
    plt.title("Recorded Profit by Region")
    plt.xlabel("Region")
    plt.ylabel("Profit")
    plt.xticks(rotation=30, ha="right")
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "profit_by_region.png", dpi=300)
    plt.close()

    # 3. Net sales by category
    category_sales = (
        df.groupby("Category")["Net_Sales"]
        .sum()
        .sort_values(ascending=False)
    )

    print("\nNET SALES BY CATEGORY")
    print(category_sales.round(2).to_string())

    print("\nCharts saved in the outputs folder.")
    print("Category and region analysis completed.")
```

```
if __name__ == "__main__":
    main()
```
