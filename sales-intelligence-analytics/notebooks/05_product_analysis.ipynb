```
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
```

```
PROJECT_DIR = Path(__file__).resolve().parent.parent
DATA_PATH = PROJECT_DIR / "data" / "Sample - Superstore.csv"
OUTPUT_DIR = PROJECT_DIR / "outputs"
```

```
def main():
    df = pd.read_csv(DATA_PATH)
    OUTPUT_DIR.mkdir(exist_ok=True)

    product_analysis = (
        df.groupby("Product_ID")
        .agg(
            Order_Count=("Order_ID", "nunique"),
            Units_Sold=("Quantity", "sum"),
            Gross_Sales=("Gross_Sales", "sum"),
            Net_Sales=("Net_Sales", "sum"),
            Recorded_Profit=("Profit", "sum"),
        )
        .sort_values("Net_Sales", ascending=False)
    )

    print("TOP 10 PRODUCTS BY NET SALES")
    print(product_analysis.head(10).round(2).to_string())

    print("\nBOTTOM 10 PRODUCTS BY NET SALES")
    print(product_analysis.tail(10).sort_values("Net_Sales").round(2).to_string())

    # Save all product-level results.
    product_analysis.to_csv(
        OUTPUT_DIR / "product_performance_analysis.csv"
    )

    # Chart: Top 10 products by net sales.
    top_products = product_analysis.head(10).sort_values("Net_Sales")

    plt.figure(figsize=(10, 6))
    top_products["Net_Sales"].plot(kind="barh")
    plt.title("Top 10 Products by Net Sales")
    plt.xlabel("Net Sales")
    plt.ylabel("Product ID")
    plt.tight_layout()
    plt.savefig(
        OUTPUT_DIR / "top_10_products_by_net_sales.png",
        dpi=300
    )
    plt.close()

    # Chart: Bottom 10 products by recorded profit.
    bottom_profit = product_analysis.nsmallest(10, "Recorded_Profit")
    bottom_profit = bottom_profit.sort_values("Recorded_Profit")

    plt.figure(figsize=(10, 6))
    bottom_profit["Recorded_Profit"].plot(kind="barh")
    plt.title("10 Products with Lowest Recorded Profit")
    plt.xlabel("Recorded Profit")
    plt.ylabel("Product ID")
    plt.tight_layout()
    plt.savefig(
        OUTPUT_DIR / "bottom_10_products_by_profit.png",
        dpi=300
    )
    plt.close()

    print("\nProduct analysis CSV and charts saved in the outputs folder.")
    print(
        "Note: Totals include all order statuses and are not verified "
        "realized sales or profit."
    )
```

```
if __name__ == "__main__":
    main()
```
